2. In which year were the highest number of TV Shows & Movies released? *(Bar Graph)*

```python
data.dtypes
```

```python
data["Date_Release"] = pd.to_datetime(data["Release_Date"],infer_datetime_format=True, errors="coerce")
```

```python
data.dtypes
```

```python
data.shape
```

```python
data_y = data["Date_Release"].dt.year.value_counts()
data_y.index, data_y.values
```

```python
data_y.plot(kind="bar")
plt.title("Number of TV Shows & Movies released by year")
plt.savefig("Figures/movie_released_by_year.png")
plt.xlabel("Year Released")
plt.show()
```
