# **Spotify Youtube Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
data = pd.read_csv('spotify_youtube_data.csv')
data.head()
```

```python
data.shape
```

```python
data.columns
```

```python
data.info()
```

```python
## Drop unwanted columns
data.drop(['Unnamed: 0','Url_spotify', 'Uri','Url_youtube'], axis=1, inplace=True)
```

```python
data.isna().sum()
```

```python
data['Likes'] = data['Likes'].fillna(0)
data['Comments'] = data['Comments'].fillna(0)
```

```python
data.isna().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data.info()
```

3. Most common **Album Types** on Spotify and their track counts  

```python
data.head()
```

```python
data["Album_type"].unique()
```

```python
data["Album_type"].value_counts()
```

```python
album_type = data["Album_type"].value_counts()
plt.pie(album_type,labels = album_type.index,autopct="%1.1f%%", startangle=60)
plt.title("Most common album types on Sportify")
plt.savefig("Figures/most_common_album.png")
plt.show()
```

4. Comparison of **average views, likes, and comments** between album types  

```python
data["Comments"].head()
```

```python
df = data.groupby("Album_type")[["Views","Likes","Comments"]].mean()
```

```python
df = df.reset_index()
```

```python
# melt - unpivot the data frame
df_melted = pd.melt(df,id_vars="Album_type",var_name="Attribute", value_name="Total")
```

```python
plt.figure(figsize=(9,10))
sns.barplot(x="Album_type", y="Total",hue="Attribute",data=df_melted)
plt.title("Average views, likes, and comments between album type")
plt.savefig("Figures/Average_views_likes_comments.png")
plt.show()
```

5. **Top 5 YouTube Channels** based on total views  

```python
data.head()
```

```python
youtube_views = data.groupby("Channel")["Views"].sum().sort_values(ascending=False).head(5)
youtube_views
```

```python
youtube_views = youtube_views.reset_index()
youtube_views
```

```python
sns.barplot(x="Views", y="Channel",data=youtube_views)
plt.title("Top 5 youtube channel")
plt.xlabel("Views")
plt.ylabel("Channel")
plt.savefig("Figures/top_5_youtube_channel.png")
plt.show()
```
